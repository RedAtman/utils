```python
import numpy as np

# Generate some example data
# np.random.seed(0)
# X = 2 * np.random.rand(100, 1)
# y = 4 + 3 * X + np.random.randn(100, 1)
# X
# y


import pandas as pd

data = {
    'Area': [50, 60, 70, 80, 90, 100],
    'Price': [150, 180, 210, 240, 270, 300]
}
df = pd.DataFrame(data)
df
```

```python
# 训练模型
# 准备数据
X = df[['Area']]
y = df['Price']
```

```python
from sklearn.linear_model import LinearRegression
# Create a linear regression model
model = LinearRegression()
model.fit(X, y)
```

```python
# 获取模型参数
# Print the model parameters
slope = model.coef_[0]
intercept = model.intercept_
print(f'Coefficient: {slope}, Intercept: {intercept}')
```

```python
# 预测新数据
# new_area = np.array([[110]])
new_area = pd.DataFrame({'Area': [110]})
predicted_price = model.predict(new_area)
print(f'Predicted price for 110 square meters: {predicted_price[0]}')
```

```python
import matplotlib.pyplot as plt
# Plot the results
# 绘制散点图和回归线
plt.scatter(X, y, color='blue', label='Data Points')
plt.plot(X, model.predict(X), color='red', label='Regression Line')
plt.xlabel('Area (sqm)')
plt.ylabel('Price (10k RMB)')
plt.title('Linear Regression')
plt.legend()
plt.show()
```
